# Task 2: SVM

## **1. Use Different Kernel and Evaluation**

### **Loading preprocessed dataset**

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Load and preprocess the data
data = pd.read_csv('preprocessed_wbdc_data.csv')
X = data.drop(columns=["Diagnosis"])
y = data["Diagnosis"]

# Split data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```


SVM classifiers use kernel functions (mathematical functions) to transform data into a higher-dimensional space, allowing them to find boundaries that separate classes. Different kernels serve different purposes:

**Linear kernel**: Works best for linearly separable data, where a straight line or hyperplane can divide the classes. It performs well when the data has minimal complexity, as it maps data directly into the feature space without adding extra dimensions.

**Polynomial kernel**: Enables more complex boundaries by mapping data to a higher dimension, useful for non-linear relationships.

**RBF (Radial Basis Function) kernel**: This is the most commonly used kernel for SVMs, particularly effective for non-linear data. And it creates flexible boundaries by mapping data with Gaussian functions. It introduces flexibility by using parameters like gamma, which controls the influence of individual data points.

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Initialize SVM models with different kernels
kernels = ['linear', 'poly', 'rbf']
svm_models = {kernel: SVC(kernel=kernel) for kernel in kernels}

# Train each SVM model and make predictions
for kernel, model in svm_models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"Kernel: {kernel}")
    print(f"Accuracy: {accuracy:.2f}")
    print("Classification Report:\n", classification_report(y_test, y_pred, target_names=['Benign', 'Malignant']))
    print("\n")

    # Compute and display confusion matrix
    conf_matrix = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(6, 4))
    sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', xticklabels=['Benign', 'Malignant'], yticklabels=['Benign', 'Malignant'])
    plt.title(f'Confusion Matrix - {kernel.capitalize()} Kernel')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()
    print("\n")
```

**Linear Kernel**:
Using the linear kernel with C=1 yields an accuracy of 0.98, with both classes achieving high precision, recall, and F1-scores around 0.97-0.98. The confusion matrix shows only 2 false positives and 2 false negatives, highlighting the linear kernel's ability to effectively separate benign and malignant cases. This strong performance suggests that the dataset is well-suited to a linear decision boundary, making the linear kernel a straightforward and effective choice.

**Polynomial Kernel**:
With parameters C=100 and degree=3, the polynomial kernel achieves an accuracy of 0.96. It performs particularly well with benign cases (recall of 0.98), though it slightly underperforms with malignant cases (recall of 0.94). The confusion matrix shows that while benign cases are classified accurately, a few malignant cases are missed. This indicates that, despite its ability to capture complex patterns, the polynomial kernel's added complexity does not significantly enhance performance over the linear kernel for this dataset.

**RBF Kernel**:
The RBF kernel reaches the highest accuracy of 0.99 with C=10 and γ=0.01. It achieves perfect precision for malignant cases and a recall of 1.00 for benign cases, demonstrating balanced sensitivity and specificity. The RBF kernel’s flexibility in modeling non-linear boundaries gives it a slight advantage, capturing subtle patterns in the data that the linear kernel might miss. This makes the RBF kernel the most effective choice among the three.

All kernels perform well, but the RBF kernel provides a slight edge, followed closely by the linear kernel. Although the polynomial kernel can model non-linear complexity, it offers limited benefits for this dataset. The linear kernel's simplicity and strong performance make it a practical option, while the RBF kernel’s adaptability to non-linear patterns provides the best overall accuracy, making it the optimal choice for this classification task.

## **2. Cross-Validation**
Then use cross-validation for hyperparameter tuning, applying GridSearchCV to optimize C and other kernel-specific parameters.

**C (Regularization)**: Controls the trade-off between a smooth decision boundary and classifying training points correctly. Higher values focus more on correct classification, which may lead to overfitting.

**Degree (Polynomial Kernel)**: Sets the polynomial degree for the kernel. Higher degrees increase model complexity but may increase the risk of overfitting.

**Gamma (RBF Kernel)**: Controls the influence of a single training sample.  Lower values create smoother, broader decision boundaries, while higher values create more complex, localized boundaries.

```python
from sklearn.model_selection import GridSearchCV

# Define hyperparameter grids for each kernel
param_grids = {
    'linear': {'C': [0.1, 1, 10, 100]},
    'poly': {'C': [0.1, 1, 10, 100], 'degree': [2, 3, 4, 5]},
    'rbf': {'C': [0.1, 1, 10, 100], 'gamma': [0.001, 0.01, 0.1, 1]}
}

# Perform Grid Search for each kernel
best_models = {}
for kernel, params in param_grids.items():
    svc = SVC(kernel=kernel)
    grid_search = GridSearchCV(svc, params, cv=5, scoring='accuracy')
    grid_search.fit(X_train, y_train)
    best_models[kernel] = grid_search.best_estimator_

    # Evaluate the best model
    y_pred = best_models[kernel].predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"Best Model for Kernel: {kernel}")
    print(f"Best Parameters: {grid_search.best_params_}")
    print(f"Accuracy: {accuracy:.2f}")
    print("Classification Report:\n", classification_report(y_test, y_pred, target_names=['Benign', 'Malignant']))
    print("\n")

```

While all kernels perform well, the RBF kernel slightly outperforms the others, followed closely by the linear kernel. The polynomial kernel, although capable of modeling complexity, doesn’t offer significant advantages for this dataset.
