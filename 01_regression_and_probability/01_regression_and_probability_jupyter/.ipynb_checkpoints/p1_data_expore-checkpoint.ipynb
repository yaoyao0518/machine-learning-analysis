## **3. Clean the data**

We use median data to fill the blank, because the median is not sensitive to outliers. In contrast to the mean, the median is less susceptible to extreme values. Especially if the distribution of the data is skewed, the median can be used to fill in the data to maintain the characteristics of the distribution.

```python
print(data.isnull().sum()) # Check for missing values

numerical_features = data.select_dtypes(include=['float64']).columns
data[numerical_features] = data[numerical_features].fillna(data[numerical_features].median())
```

```python
import numpy as np
# Create a new feature: average number of rooms per household
data['rooms_per_household'] = data['total_rooms'] / data['households']
data['bedrooms_per_poupulation'] = data['total_bedrooms'] / data['population']
data['longitude_divide_latitude'] = data['longitude'] / data['latitude']

# Removal of multicollinearity and features
data = data.drop(columns=['total_rooms', 'total_bedrooms', 'population', 'households', 'longitude', 'latitude'])
```

```python
# one-hot encoding
data_encoded = pd.get_dummies(data, columns=['ocean_proximity'], dtype = int)
data_encoded
```

**Remove outliers**

Outliers were removed using the Interquartile Range (IQR) method which is effective for handling extreme values in skewed data distributions.

```python
import pandas as pd

numerical_features = data_encoded.select_dtypes(include=['float64']).columns

# Detect and remove outliers using the IQR method
def remove_outliers_iqr(df, features):
    for feature in features:
        Q1 = df[feature].quantile(0.25)
        Q3 = df[feature].quantile(0.75)
        IQR = Q3 - Q1

        # lower and upper bounds for outliers
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        # Remove outliers
        df = df[(df[feature] >= lower_bound) & (df[feature] <= upper_bound)]

    return df

data_cleaned = remove_outliers_iqr(data_encoded, numerical_features)

# Display the number of rows before and after outlier removal
print(f"Original data shape: {data_encoded.shape}")
print(f"Cleaned data shape: {data_cleaned.shape}")

data_cleaned.to_csv("/content/drive/MyDrive/A1_Classification/data/cleaned_data.csv", index = False) # When saving the DataFrame to a CSV, always use index=False to avoid saving the index as a column in the CSV file
data_cleaned
```

```python
data_cleaned.hist(bins=30, figsize=(15, 10))
plt.suptitle('Histograms for Variables', fontsize=16)
plt.show()
```

After processing, the variables are in much better shape then before, but there are still some skewed distribution. For the categorical feature ocean_proximity, one-hot encoding has been successfully applied, though the imbalance in the ISLAND category remains noticeable, with very few instances. This imbalance may still pose challenges during modeling, especially in accurately predicting underrepresented categories.
